```python
from datasets import load_dataset

ds = load_dataset("openlifescienceai/medmcqa")
# Display the first 5 rows of the training set
import pandas as pd

df = pd.DataFrame(ds['train'])
df.head(5)
```

```python
# Add correct_answer column to the original dataframe
df['correct_answer'] = df.apply(lambda row: row['opa'] if row['cop'] == 0 
                                           else row['opb'] if row['cop'] == 1 
                                           else row['opc'] if row['cop'] == 2 
                                           else row['opd'] if row['cop'] == 3 
                                           else None, axis=1)

# Add complete_answer column that joins correct_answer and exp
df['complete_answer'] = df['correct_answer'] + ' | ' + df['exp']

# Add complete_question column that joins question and all options
df['complete_question'] = df['question'] + ' | Options: ' + df['opa'] + ', ' + df['opb'] + ', ' + df['opc'] + ', ' + df['opd']

# Add question_with_context column that joins subject, topic, and complete question
df['question_with_context'] = 'Subject: ' + df['subject_name'].fillna('') + ' | Topic: ' + df['topic_name'].fillna('N/A') + ' | Question: ' + df['complete_question']

# Display first few rows with all columns
df.head()
```
